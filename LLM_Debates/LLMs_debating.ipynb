```python
import ollama
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key = 'dummy_key'
)
```

```python
model_judge= 'deepseek-r1:7b'
model_debater_A= 'llama3.2:latest'
model_debater_B = 'gemma3:4b'



system_prompt_judge = 'You are a judge in a debate. You may be asked either to generate a debate topic or to evaluate a participant\'s response. Be funny at all times'
system_prompt_debater = 'You are in a debate. Be funny at all times. Write exactly two witty but logically correct arguments. '
```

```python

messages_judge=[{
  'role':'system',
  'content':system_prompt_judge },
  { 'role':'user',
  'content':

f"""Generate one original, funny but debatable question.

The question should be amusing, grammatically correct, and easy to understand and it must allow two intelligent people to argue convincingly for opposite sides.
Examples you can learn from:
- Should pineapple belong on pizza?
- Should cats be allowed to drive cars?
- Should cereal be eaten with a fork?
- Should socks come in a pack of three?
Do NOT closely imitate the examples.
Generate 5 funny debatable topics.
Choose the FUNNIEST ONE that is also genuinely DEBATABLE.
Return only the ONE CHOSEN topic.
Only respond with the topic in one sentence, nothing else"""}]

response_judge = client.chat.completions.create(
    model=model_judge,
    messages=messages_judge)
debate_topic=response_judge.choices[0].message.content
display(Markdown(debate_topic))
```

```python
display(print(response_judge))
```

```python
messages_A=[{'role': 'system', 'content': system_prompt_debater},
                     {'role':'user',
                      'content': f"""Debate STRICTLY IN FAVOR of this topic:
                      {debate_topic}
                      give no more than two arguments and two sentences for each argument""" }]

response_A=client.chat.completions.create(
                    model=model_debater_A,
                    messages=messages_A,
                    max_tokens=200)
response_A=response_A.choices[0].message.content

display(Markdown(response_A))
```

```python

messages_B= [{'role':'system', 'content': system_prompt_debater},
                   {'role':'user','content':f"""Debate STRICTLY AGAINST this topic:
                                           {debate_topic}
                                          Your opponent's argument is:
                                           {response_A},
                                          Respond to and refute your opponent's points
                                          give no more than two arguments and two sentences for each argument""" }]

response_B=client.chat.completions.create(
    model=model_debater_B,
    messages=messages_B,
    max_tokens=200)
# print("=====",response_B.choices[0].finish_reason)
# print("=====",response_B.choices[0].message.reasoning)
# print("=====",response_B)

# print(repr(response_B.choices[0].message.content))
response_B = response_B.choices[0].message.content
display(Markdown(response_B))
```

```python

messages_judge = [
    {'role': 'system','content': system_prompt_judge},
    {'role':'user', 'content': f""" Debate topic:
                                    {debate_topic}
                                    Debater A's response:
                                     {response_A}
                                    Debater B's response:
                                     {response_B}

                                    Compare both responses.
                                    Score each out of 10.
                                    only give one line of feedback per debater.
                                    Declare the winner."""}]

response_judge = client.chat.completions.create(
    model=model_judge,
    messages=messages_judge)
judge_feedback=response_judge.choices[0].message.content

display(Markdown(judge_feedback))
```
