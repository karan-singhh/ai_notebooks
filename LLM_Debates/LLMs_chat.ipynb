```python
import ollama
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key = 'dummy_key'
)
```

```python
model_judge= 'llama3.2:1b'
model_debater= 'deepseek-r1:1.5b'

system_prompt_judge = 'You are a judge in a debate. You may be asked either to generate a debate topic or to evaluate a participant\'s response. Be funny at all times'
system_prompt_debater = 'You are in a debate. Be funny at all times. Write exactly two witty but logically correct arguments. '
```

```python

messages_judge=[{
  'role':'system',
  'content':system_prompt_judge },
  { 'role':'user',
  'content':'Generate a funny sounding topic for this debate. Only answer in one sentence with just the topic, nothing else.'}]
response_judge = client.chat.completions.create(
    model=model_judge,
    messages=messages_judge)
debate_topic=response_judge.choices[0].message.content
print(debate_topic)
```

```python

messages_debater= [{'role':'system', 'content': system_prompt_debater},
                   {'role':'user','content':f'Debate on this topic: {debate_topic} '}]
response_debater=client.chat.completions.create(
    model=model_debater,
    messages=messages_debater)
debater_arg = response_debater.choices[0].message.content

# print(response_debater.choices[0].message.content)
```

```python

messages_judge = [
    {'role': 'system','content': system_prompt_judge},
    {'role':'user', 'content': f""" Debate topic: {debate_topic}
                                    Participant's response: {response_debater}
                                    Judge the participant/debater\'s response, give score out of 10, and provide feedback. Be funny"""}]

response_judge = client.chat.completions.create(
    model=model_judge,
    messages=messages_judge)
judge_feedback=response_judge.choices[0].message.content

display(Markdown(judge_feedback))
```
